```python
import matplotlib.pyplot as plt
import numpy as np
import PIL.Image as Image
import os
import textwrap
```

```python
fix, ax = plt.subplots(1, 3, figsize=(12, 4))
captions = [
    "a Miyazaki-like illustration of a traveling man with a warm smile, gentle color gradients",
    "a Ghibli-style close-up of a playful girl with wind-blown hair, soft ambient light",
    "a Ghibli-style close-up of a curious child with tousled hair, dreamy background"
]

for i in range(3):
    img = Image.open(os.path.join("images", f"playaround_large{i+1}.png"))
    ax[i].imshow(np.array(img))
    wrapped = textwrap.fill(captions[i], width=40)
    ax[i].set_xlabel(wrapped, fontsize=12)
    ax[i].set_frame_on(False)
    ax[i].set_xticks([])
    ax[i].set_yticks([])

plt.tight_layout()
plt.show()
```
